### DataFrame() Двумерный массив данных
DataFrame() - это класс который создаёт двумерные таблицы (строки и колонки) по оси 0 и оси 1. Основной объект Pandas.

Берётся класс из библиотеки pd
DataFrame обычно называют df

pd.DataFrame(data, index, columns, dtype, copy) = Парматеры класса
- data	Данные (словарь, список, массив)	{'A': [1, 2], 'B': [3, 4]}
- index	Метки строк	['row1', 'row2']
- columns	Названия колонок	['A', 'B']
- dtype	это принудительная конвертация данных в нужный тип. 
- copy	Копировать данные?	True / False

Параметр data:
Класс принимает 3 типы данных:
- Словарь
- Список
- NumPy масссив



#### date = dict

```python
# Словарь со сколярным значениями
# В словарях указывать пармерт columns не нужно указывать так как у нас ключи это колонки
# В словарях обычно не указывают явный тип данных через dtype так как у нас в разных колонках разные типы данных

df = pd.DataFrame({
    'name': 'Umed', # Одно значений (сколярное значение)
    'age': 22,      # Одно значений (сколярное значение) 
},  index=['0'])     # Без index pandas не знает, как назвать единственную строку.
                    # Названием колонок становяться ключи словаря 
df
```

```python
# Словарь со значениями список
# В словарях указывать пармерт columns не нужно указывать так как у нас ключи это колонки
# В словарях обычно не указывают явный тип данных через dtype так как у нас в разных колонках разные типы данных
df2 = pd.DataFrame({
    'name': ['UMED', 'Numon'], # Если значене ключа у нас список из N значений то у нас инедксы (index) строк исоздаются сами  
    'age': [20, 25]            # Если значене ключа у нас список из N значений то у нас инедксы (index) строк исоздаются сами
})                             # Названием колонок становяться ключи словаря
df2
```

```python
# Данные редко указывают внутри класса, 
# обычно их присваивают перменным а потом передают в качестве значений параметров

# Задаём переменные 
names = ['Umed', 'Numon', 'Ali', 'Numon']
ages = [20, 30, 18, 22]

# Присваиваем перменные в качестве занченйи ключей нашего словаря
df = pd.DataFrame({
    'names': names,
    'ages': ages,
})
df
```

### Series()
Series — это класс, который создаёт одномерный массив с индексом. Как один столбец таблицы.

Берётся класс из библиотеки pd

pd.Series(data, index, dtype, name, copy)

- data	Данные	[1, 2, 3]
- index	Метки	['a', 'b', 'c'] Присваиваем инидексы явно через парметр index
- dtype	Тип данных	= int64 это принудительная конвертация данных в нужный тип. 
- name	Имя Series	'my_series'
- copy	Копировать?	True / False

```python
# Создаём одну колонку класс Series()
# Явна присваивать индексы не тнеобходимости 
df = pd.Series([10, 20, 30])
df
```

```python
# Также мы можем явно указывать значения индексов
df = pd.Series([1,2,3], 
               index=['a','b','c'] # Явна Присваиваем инедксы в виде букв
              )
df
```

```python
# Указываем индексы класса в виде словаря, где ключь у нас индекс а значение его это цифра
# Ключь это индекс, 
# Значение ключа это значение колонки 

df = pd.Series({'a': 1, 'b':2, 'c':3}) 
df
```

```python
# Можно также явна укзаать имя Series()
df = pd.Series(
    [10,10,10],
    index=['a','b','c'],
    name='value', # Указываем на имя Series
)
df
```

```python
# Также монжо сразу явно конвертировать типы данных Serie
df = pd.Series(
    [10,10,10],
    index=['a','b','c'],
    name='value', 
    dtype=float, # Явна меняем тип данных 
)
df
```

```python
# В Serias также можн передавать перменные в качесве аргументов параметров

# Создаём перменную которую передадим в класс Serias
num01 = [10,11,12]

df = pd.Series(num01) # Передаём в качестве параметра в парметр data перменную содержащую массив данных
df
```

### read_csv() = чтение csv файла. 
pd.read_csv(filepath, sep, header, names, index_col, usecols, dtype, encoding, ...)
Основные параметры
Параметр	 Что означает	Пример
- filepath  - Путь к файлу	- 'data.csv'
- sep	    - Разделитель	- ',', ';', '\t'
- header    - Строка с заголовками	- 0 (первая)
- names	    - Свои названия колонок	- ['A', 'B']
- index_col - Колонка-индекс	    - 'id'
- usecols	- Какие колонки читать	- ['A', 'B']
- dtype	    - Типы данных	        - {'A': int}
- encoding	- Кодировка	            - 'utf-8'
- nrows	    - Сколько строк читать	- 100
- skiprows	- Пропустить строк	    - 5
- na_values	- Что считать NaN	    - ['NA', '']

```python
# Читаем по абсолютному пути
# Cчитает первую строку файла заголовком (header), а не данными.
df = pd.read_csv('E:/python/notebooks/Pandas/sandb.csv')
df.head() 
```

```python
# Читаем по названию файла, 
# через такой метод мы можем считать только те файлы которые находяться в той директории
# с которой работаем 
# Cчитает первую строку файла заголовком (header), а не данными.
df = pd.read_csv('sandb.csv')
df.head()
# read_csv() читает файл с первой строки, но первую строку по умолчанию использует как заголовок, а не как данные.
#  Реальные данные начинаются со второй строки файла (если header не отключён).
```

```python
# header  = указываем строку которая становится загаловком в DataFrame, всё что после этой строки это данные 
# nrows = указываем сколько строк читать 
df  = pd.read_csv('E:/python/notebooks/Pandas/sandb.csv',
                  header=3, # заголовок на 3-й строке (0,1,2 → 2-я), по умолчанию она 0 
                  nrows=10) # Сколько строк считывать после загаловка 
df
```

```python
# index_col = «возьми вот эту колонку и сделай её индексом (подписями строк) вместо стандартных 0, 1, 2, 3...».
df  = pd.read_csv('E:/python/notebooks/Pandas/sandb.csv',
                  header=3, # заголовок на 3-й строке (0,1,2 → 2-я), по умолчанию она 0 
                  nrows=10, # Сколько строк считывать после загаловка 
                  index_col='fpd_01_shis')
df
```

```python
# usecols = Говорит pandas: читай только эти колонки, остальные игнорируй (даже не загружай их в память).
df  = pd.read_csv('E:/python/notebooks/Pandas/sandb.csv',
                  header=3, # заголовок на 3-й строке (0,1,2 → 2-я), по умолчанию она 0
                  nrows=10, # Сколько строк считывать после загаловка 
                  usecols=['fpd_01_shis', 'cum_znam_01']) # Отбирает толкь оте колонки которые указать 
df
```

```python
# dtype = мы говорим pandas, каким типом читать колонку с самого начала, а не меняем тип уже после загрузки.
# names = names заменяет заголовок, но при этом первая строка файла остаётся как данные 
# usecols = Говорит pandas: читай только эти колонки, остальные игнорируй (даже не загружай их в память).
df  = pd.read_csv('E:/python/notebooks/Pandas/sandb.csv',
                  header=0, # заголовок на 3-й строке (0,1,2 → 2-я), по умолчанию она 0
                  nrows=10, # Сколько строк считывать после загаловка 
                  names=['A','B'],   # names заменяет заголовок, но при этом первая строка файла остаётся как данные 
                  dtype={'A':str})   # Колонку которая имеет заголовок A читает как str 
df
```

```python
# skiprows — пропустить строки сверху
# Говорит pandas: не читай первые N строк файла, начни сразу с нужной.
# С skiprows=5 первые 5 строк (весь «мусор» сверху) пропускаются, и pandas начинает читать

df  = pd.read_csv('E:/python/notebooks/Pandas/sandb.csv',
                  header=0, # заголовок на 3-й строке (0,1,2 → 2-я), по умолчанию она 0
                  nrows=10, # Сколько строк считывать после загаловка 
                  names=['A','B'],   # names заменяет заголовок, но при этом первая строка файла остаётся как данные 
                  dtype={'A':str},   # Колонку которая имеет заголовок A читает как str 
                  skiprows=2)        # Пропускае тсверзу то кол-во строк которое указать 
df
```

### read_excel() = чтение excel файла.
Читает Excel-файл (.xlsx, .xls) → DataFrame.

Синтаксис

pd.read_excel(io, sheet_name, header, names, index_col, usecols, dtype, ...)

Основные параметры

Параметр	Что означает	Пример
- io	Путь к файлу	'data.xlsx'
- sheet_name	Имя/номер листа	'Sheet1', 0, None
- header	Строка с заголовками	0
- names	Свои названия	['A', 'B']
- index_col	Колонка-индекс	'id'
- usecols	Какие колонки	'A:C'
- dtype	Типы	{'A': int}
- skiprows	Пропустить строк	5
- nrows	Сколько строк	100

### read_json() =  чтение json файла.
Синтаксис

pd.read_json(path_or_buf, orient, dtype, convert_axes, ...)

Основные параметры :
- Параметр	Что означает	Пример
- path_or_buf	Путь к файлу	'data.json'
- orient	Формат JSON	'records', 'columns', 'index'
- dtype	Типы	{'A': int}
- convert_axes	Преобразовывать оси	True / False
- lines	Читать построчно	True / False

Форматы orient:
- orient	Формат JSON
- 'records'	[{"A": 1, "B": 2}, ...]
- 'columns'	{"A": {"0": 1}, "B": {"0": 2}}
- 'index'	{"0": {"A": 1, "B": 2}}
- 'values'	[[1, 2], [3, 4]]

### read_sql() = чтение из БД.

Читает данные из SQL-базы → DataFrame.

Синтаксис

pd.read_sql(sql, con, index_col, columns, ...)

Основные параметры:

- con	Соединение с БД	sqlite3.connect('db.db')
- index_col	Колонка-индекс	'id'
- columns	Колонки	['name', 'age']

# Выбор и фильтрация
- При создании DataFrame важно учитывать, что для неё автоматически создаются метки это её инедксы
- Но инедксы мы можем задавать и сами
: = указыват срез откуда до кудова отбирать строк либо колонки
['a':'c'] = отбираем строки с метки a по с тоесть a,b,c
: [:] отбирает все строки 

Если мы задёпм метки сами и мы хотим обратиться по этим меткам то мы можем воспользоватся методом
- df.loc[] — выборка по меткам (label-based)
loc работает с именами индекса и именами столбцов. Верхняя граница включается (в отличие от обычного Python-слайсинга).

### loc[]
по меткам

```python
df = pd.DataFrame({
    'name':['Alice','Bob','Carol','Dave'],
    'age':[30,25,35,28],
    'city':['NY','LA','NY','SF']
}, index = ['a','b','c','d'])
df
```

```python
# Берём из DataFrame строку по метке которой мы сами создали по метке 'b'
# df.loc[строки, столбцы] = синтаксис
df.loc['b'] # Вызов одной строки по метке
```

```python
# Выбираем строки и колонки 
df.loc[['a', 'b'], 
    ['name', 'age']] 
```

```python
df.loc[['a', 'b']]# Вызов двух строк
```

```python
df.loc['a':'c'] # Cрез с метками 
```

```python
# Все строки по каким колонкам
df.loc[:,'name':'age']
```

```python
# Все колонки и 2 строки
df.loc[['a', 'b'], :]
```

Разберём группировку и агрегацию — это одна из самых мощных тем в pandas. Здесь логика такая: разбить → применить → собрать.

```python
df = pd.read_excel('E:/python/notebooks/Pandas/sandb_Excel.xlsx')
df
```

```python

# Задаём переменные 
names = ['Umed', 'Numon', 'Ali', 'Numon']
ages = [20, 30, 18, 22]

# Присваиваем перменные в качестве занченйи ключей нашего словаря
df = pd.DataFrame({
    'names': names,
    'ages': ages,
})
df
```

```python
# df.groupby() — разбить на группы
# Говорит pandas: сгруппируй строки по значению в колонке.
df.groupby('names')
# Само по себе это ничего не покажет — вернётся объект DataFrameGroupBy, а не таблица. Нужно применить агрегацию.
```

```python
# groupby возвращает объект, который можно передат дальше и посчитать агрегированные показатели по всем остальным колонкам
# Например средню по каждой группе 

df.groupby('names').mean()
```

```python
# Сумму по каждой группе 
df.groupby('names').sum()
# Зачем: суммарные продажи, общая выручка по региону и т.д.
```

```python
# Кол - во сущьностей в каждой группе 
df.groupby('names').count()
# Зачем нужно : сколько записей в каждой группе.  count() не считает NaN — считает только непустые.
```

```python
df.groupby('names').min()
# показывает минимальный элемент по каждой группе
```

```python
df.groupby('names').max()
# показывает максимальный элемент по каждой группе 
```

```python
#.agg() — несколько агрегаций сразу
# Это главный метод, когда нужно посчитать несколько функций одновременно или разные функции для разных колонок.
# Необходимо отдельно указывать по какой колонке предстоит считать все показатели в agg()
# Если необходимо посчтать по колонке несколько показателей, то необходиимо сделат это в [] списке
df.groupby('names')['ages'].agg(['mean','min','max','count', 'sum'])
```

## Roll Rate

```python
import pandas as pd
!pip install seaborn
import seaborn as sns


# Long format: одна строка = один кредит × один месяц
df = pd.DataFrame({
    'loan_id':    [100001, 100001, 100001, 100001,
                   100002, 100002, 100002,
                   100003, 100003, 100003],
    'mob':        [1, 2, 3, 4,
                   1, 2, 3,
                   1, 2, 3],
    'dpd_bucket': ['Current', 'Current', '1-30', '31-60',
                   'Current', '1-30', 'Current',
                   '1-30', '31-60', '61-90']
})

print(df)

df = df.sort_values(['loan_id', 'mob']).reset_index(drop=True)


# Текущее состояние (месяц t)
from_state = df[['loan_id', 'mob', 'dpd_bucket']].rename(
    columns={'dpd_bucket': 'from_bucket'}
)

# Следующее состояние (месяц t+1)
to_state = df[['loan_id', 'mob', 'dpd_bucket']].copy()
to_state['mob'] = to_state['mob'] - 1  # сдвигаем на 1 назад
to_state = to_state.rename(columns={'dpd_bucket': 'to_bucket'})

# Объединяем по loan_id и mob
pairs = from_state.merge(to_state, on=['loan_id', 'mob'], how='inner')



print(pairs)


# Упорядочим бакеты
order = ['Current', '1-30', '31-60', '61-90', '90+']

# Строим матрицу переходов
roll_matrix = pd.crosstab(
    pairs['from_bucket'],
    pairs['to_bucket'],
    normalize='index'  # проценты по строкам
) * 100

# Приводим к нужному порядку
roll_matrix = roll_matrix.reindex(index=order, columns=order, fill_value=0)

print(roll_matrix.round(2))

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.heatmap(roll_matrix, annot=True, fmt='.1f', cmap='RdYlGn_r')
plt.xlabel('TO (Куда перешёл)')
plt.ylabel('FROM (Откуда пришёл)')
plt.title('Roll Rate Matrix (%)')
plt.show()
```
