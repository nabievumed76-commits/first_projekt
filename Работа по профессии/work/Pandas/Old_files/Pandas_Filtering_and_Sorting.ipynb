```python
import pandas as pd
import numpy as np 
```

```python
# Задача создать табличку с данными 
# Для этого используем словарь
forum_user = {
    'User_ID': [1,2,3,4,5],
    'Username':['umed','namt','numon','amid','help'],
    'Age':[18,42,54,33,None],
    'Joined_Date':pd.to_datetime(['2032-09-02','2032-05-04','2032-06-03','2032-12-10', '2032-11-28']),
    'Total_Post': [150,230,80,420,310],
    'Reputation':[500,760,200,950,500]
}

# Тепер из словаря нужно сделать таблицу для этого есть класс DateFrame из пакета pandas
df = pd.DataFrame(forum_user) # Создаём таблицу и называвем её df
df
```

### Filter

```python
df['Age']>= 25 
# Получаем серию в которая значения тех строк которые боьше либо равны 25 будут становиться True
```

```python
# Пример фильтрации по строкам
df[df['Age'] >= 25] # Конвертируем серию в DataFrame
# все строки которые не убовлетворяют условие не будт оотображены в итоговой таблицу
```

```python
df[df['Total_Post']>= 200]
```

### Операторы & и | двойные условия 

```python
# оператор и &
# Сначала применяестя фильтрация по одной колонке потом по второй колонке 
# После фильтрации убираются строки по обоим колонкам и выводится конечная таблица 
df[(df['Age']>=25) & (df['Total_Post']>= 100)] # Двйоноеусловие срабатывает только в случаи использования оператора и &
# Двоное условие 
```

```python
# оператор и |
df[(df['Age']>=25) | (df['Total_Post']>= 100)]
```

```python
# бёрм те строки из колонки которые равны определёному значению
df['Reputation'].isin([200,500]) # метод .isin() оставляет только те строки                                                                                                          
```

### isin() and range(), date_range()

```python
df[df['Reputation'].isin([200,500])] # метод .isin()
# Метод отбирает тошльок те строки из колонки которые равны 200 либо 500
```

```python
df[df['Reputation'].isin(range(200, 500))] 
# берутся только те строки из колонки которые начинаются с 200 и заканчиваются на 400
```

```python
date_range =pd.date_range(start='2032-03-01', end='2032-08-01')
date_range
```

```python
df[df['Joined_Date'].isin(date_range)] 
# отбираем те строки из колонки занчение которыч равны в диапазоне указаных дат  
```

```python
df[df['Joined_Date'].isin(pd.date_range(start='2032-03-01', end='2032-08-01'))]
```

## Sorting

```python
df.sort_values(by='User_ID', ascending=True) # метод сортирует по возрастанию
# Метод фиьтрует по колонке Age по возрастанию
# Сортировка по колонке
```

```python
# Сортировка по строке
df[['Age', 'Total_Post', 'Reputation']].sort_values(by=2, axis=1 ,ascending=False)
# b=2 соритровка по инедксу 2 
# axis=1 сортировка по колонкам 
# ascending=False сортировка по убыванию
```
