```python
import pandas as pd 
# 1. Чтение и запись данных
# Загрузка из CSV, Excel, JSON, SQL, Parquet и др.
# 2. Структуры данных 
# Series — одномерный массив с метками (как колонка)
#   DataFrame — двумерная таблица (строки × столбцы), главный объект
```

## DataFrame

```python
# Задача создать табличку с данными 
# Для этого используем словарь
forum_user = {
    'User_ID': [1,2,3,4,5],
    'Username':['umed','namt','numon','amid','help'],
    'Age':[18,42,54,33,None],
    'Joined_Date':pd.to_datetime(['2032-09-02','2032-05-04','2032-06-03','2032-12-10', '2032-11-28']),
    'Total_Post': [150,230,80,420,310],
    'Reputation':[500,760,200,950,500]
}

# Тепер из словаря нужно сделать таблицу для этого есть класс DateFrame из пакета pandas
df = pd.DataFrame(forum_user) # Создаём таблицу и называвем её df
df
```

```python
df.describe() # Метод
# Метод, который считает основные статистики по каждой числовой колонке 
# и возвращает результат в виде нового DataFrame.
# Колонки из строк, дат и объектов не входят в итоговый DataFrame
```

```python
type(df.describe()) # Смотрим тип данных который возвращает метод 
```

```python
df.describe().index # Через атрибут .index наложеный на метод .descibe(), который показывает инедксы DataFrame 
# который получаем в итоге
# describe() создаёт новую таблицу - Получается новый DataFrame (не тот же df!)
# .index показывает метки этого нового DataFrame
```

```python
type(df.describe().index)
```

```python
df.describe().round(2) 
# Цепочка методов. 
# Метод .describe() принимает объект df и выводит новый объект DataFrame
# Далее по цепочке идёт другой метод .round(), который принимает новый объект DataFrame 
# и округляет до 2 знаков после запятой
```
