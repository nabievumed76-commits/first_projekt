```python
import pandas as pd
import numpy as np 
```

```python
# Задача создать табличку с данными 
# Для этого используем словарь
forum_user = {
    'User_ID': [1,2,3,4,5],
    'Username':['umed','namt','numon','amid','help'],
    'Age':[18,42,54,33,None],
    'Joined_Date':pd.to_datetime(['2032-09-02','2032-05-04','2032-06-03','2032-12-10', '2032-11-28']),
    'Total_Post': [150,230,80,420,310],
    'Reputation':[500,760,200,950,500]
}

# Тепер из словаря нужно сделать таблицу для этого есть класс DateFrame из пакета pandas
df = pd.DataFrame(forum_user) # Создаём таблицу и называвем её df
df
```

## Merge слияние Таблиц

```python
total_comments = {
    'Total_Comments': [20,30,40,50,60,70]
}


df2 = pd.DataFrame(total_comments)
df2
```

```python
pd.concat([df, df2]) # метод concat([]) слияние 2 таблиц
# По умолячанию слоияние происходит по оси 0 
```

```python
df3 = pd.concat([df, df2], axis=1)
# Указываем сами по какой оси слиять (по ои axis=1)
# Добавялется новая колонка 
df3
```

## .drop Удаление колонок либо строк

### Удаление колонок

```python
# df.drop(['Language']) 
# По умолчанию метод удаляет по оси 0 axis=0 тоесть по строкам, а мы указываем колонку, 
# поэтому необходимо менят ось на 1 
```

```python
df.drop(['Language'], axis=1)
# Оригинальная табличка не меняется для этого небхо
```

```python
df #оригинальная таблица при этом не меняется для его изменения необходимо указать параметр inplace=True
```

```python
df.drop(['Language'], axis=1, inplace=True)
# Теперь оригинальная таблица изменилась, колонка Lаnguage удалилась 
df
```
