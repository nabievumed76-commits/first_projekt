```python

# СПИСОК ПАР (fpd, cum_debt, delta)

pairs = [
    ('fpd_01_befor', 'cum_debt_01_befor', 'delta_01_befor'),
    ('fpd_01_after', 'cum_debt_01_after', 'delta_01_after'),
    ('fpd_05_befor', 'cum_debt_02_befor', 'delta_05_befor'),
    ('fpd_05_after', 'cum_debt_02_after', 'delta_05_after'),
    ('fpd_07_befor', 'cum_debt_03_befor', 'delta_07_befor'),
    ('fpd_07_after', 'cum_debt_03_after', 'delta_07_after'),
]


# ЦИКЛ ПО ПАРАМ

for fpd_col, debt_col, delta_col in pairs:
    df[delta_col] = df[fpd_col].cumsum() / df[debt_col].cumsum()

df
```
