# Функция arange()  — это функция модуля NumPy, которая создаёт массив с последовательностью чисел в заданном диапазоне с заданным шагом.

```python
import numpy as np
```

```python
np.arange(10) # Создаём упорядоченый одномерный массив numpy
```

```python
np.arange(15,25) # Создаём упорядоченый одномерный массив numpy от одного  значения до дургово
```

```python
np.arange(10,25, 5) # Создаём упорядоченый одномерный массив numpy от одного  значения до другово 
# c определёным шагом 5
```

```python
np.arange(-10, 30, 2) # Создаём упорядоченый одномерный массив numpy от одного  значения до другово 
# c определёным шагом 2
```

# Функция reshape() — это метод массива NumPy, который меняет форму массива, не меняя сами данные.

```python
np.arange(20).reshape([2,10]) # Из массива с 20 элеменатми создаём двумерный массив
```

```python
np.arange(20).reshape([4,5]) # в reshape указываем форму массива 
```

```python
np.arange(20).reshape([4,5]).reshape([2,2,5]).reshape([4,5]) # Reshape всего лишь меняет форму массива
# Но можно также менять форму в большую мерность и меньшую мерность, всё в одной строке.  
```

```python
np.random.random((10,)).reshape([2,5])
```

```python
np.random.seed(10)
np.random.random((4,10)).reshape([2,20]) 
```

# Flaten - это метод массива NumPy, который превращает многомерный массив в одномерный, копируя данные.

```python
np.random.seed(10)
np.random.randint(10, 80, (4,3,3))
```

```python
np.random.seed(10)
np.random.randint(10, 80, (4,3,3)).flatten() # Меняем трёх мерный массив в одномерный 
```
