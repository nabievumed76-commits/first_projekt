## 1. Calculating Total Revenue

```python
prices = np.array([19.99,29.99,14.99,9.99,24.99]) # Цены товаров (колонка цен)
```

```python
print(prices)
```

```python
print(quan)
```

```python
print(revenue) 
```

```python
total_rev = np.sum(revenue)
```

```python
print(total_rev) # Общая сумма выручки
```

## 3. Splitting Order into Batches

```python
order_ids = np.array([1001, 1002, 1003,1004, 1005, 1006, 1007, 1008, 1009,1010,1011,1012])
print(order_ids)
```

```python
batchas = np.split(order_ids, 4) # Разделяем на батчи, 
# указываем конкрктено на сколько батчей разделить 
print(batchas)
```

```python
for bath in batchas:
    print(bath)
```

```python
for bath in batchas: # берём элемент по индексу 0 
    print(bath[0])
```

```python
for i, bath in enumerate(batchas):
    print(f'Batch number {i}:')
    print('First element in the batch:', bath[0])
    print('Second element in the bath:', bath[1])
```

## 4. Categorizing Product Rating

```python
ratings = np.array([4.5,3.2,2.8,5.0,4.1,3.9,2.5,4.7])
print(ratings)
```

```python
positiv_ratings = ratings[ratings >= 4.0] # Используя квадратные скобки мы можем 
# отфильтровать объекты в массиве numpy.
# Здесь находим позитивные отзывы
print(positiv_ratings)
```

```python
negative_ratings = ratings[ratings <= 4.0]
# Здесь находим негативные отзывы
print(negative_ratings)
```

## 7 Generation of the sample stock data

```python
# Создаём данные 
# Из этих данных будет создан массив
companies = ['Google', 'Microsoft', 'Apple'] 
days = ['Mon 1 April', 'Thu 2 April']
price_types = ['Open', 'Close', 'High', 'Low']
```

```python
np.random.seed(1) # Задаём стартовое число для генератора случайных цифр
# Далее у нас создаются случайные даные 
#  Ось 0 — это «слои» (глубина), по одному на каждую компанию.
#  Ось 1 — это строки, по одной на каждый день.
#  Ось 2 — это столбцы, по одному на каждый тип цены.
stock_prices = np.round(np.random.random((len(companies), len(days), len(price_types))), 3)
# np.random.random(shape) ждёт числа (или кортеж чисел) — размеры будущего массива.
# len(companies) = 3, len(days) = 2, len(price_types) = 4 — это числа, которые random() и ждёт.
# Получаем кортеж (3, 2, 4), и random() создаёт массив этой формы со случайными числами.
print(stock_prices)
```

```python
stock_prices.shape # Форма нашего массива
```

```python
for i, company in enumerate(companies):
    print(f'Stock prices for the {company}:')
    for j, day in enumerate(days):
        print(f'Day:{day}')
        for k, price_type in enumerate(price_types):
            print(f'{price_type} Price: {stock_prices[i,j,k]}')
        print('')
    print('')
    
```
