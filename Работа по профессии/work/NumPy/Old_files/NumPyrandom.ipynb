# Создание массиво с помощю функций

```python
import numpy as np
```

```python
np.zeros((3,4,2))  #  .zeros() это функция NumPy, которая создаёт массив, заполненный нулями, 
# по умолчанию float
```

```python
np.zeros((3,4),dtype=int) # Создаём массив 2 мерный с 3 строками и 4 колонками, типа int
```

```python
np.ones((5,3),dtype=int) # это функция NumPy, которая создаёт массив, заполненный единицами.
```

```python
np.random.random((2,4)) # Создай массив из рандомных чисел
# np	NumPy	модуль (пакет)
# np.random	подмодуль внутри NumPy	модуль
# np.random.random	функция внутри подмодуля	функция
# Диапазон: каждое число от 0.0 до 1.0 (не включая 1.0)
```

```python
np.random.seed(10)   # ed — это функция модуля NumPy, 
# Которая задаёт стартовое число (seed) для генератора псевдослучайных чисел.
np.random.random((2,4)) # Создай массив из рандомных чисел которые начинаются с .seed(10)
# np	NumPy	модуль (пакет)
# np.random	подмодуль внутри NumPy	модуль
# np.random.random	функция внутри подмодуля	функция
# Диапазон: каждое число от 0.0 до 1.0 (не включая 1.0)
```

```python
np.random.seed(15)
np.random.randint(10,20, size=(2,4)) # В начале укказываем чсило с которого 
# Необходиомо выбирать случайные числа 
# Потом указываем до какой границы брать числа
# Далее указываем какой массив создать 
```

```python
np.random.seed(10)
np.random.uniform(1, 10, size=(2,4)) # "Эта фуцкнция работате как random но здесь можнo указать диапазон, 
# данные будут по типу float  
```

```python
np.random.choice(np.array([1,2,3]), size=20) 
```
